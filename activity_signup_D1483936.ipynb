## 題目說明

本程式整理「桌遊體驗」與「電影欣賞」兩場活動的報名名單。原始名單是以逗號分隔的文字，裡面有多餘空白、大小寫不一致，也有同一人重複報名的情況。

程式先移除空白、統一轉成小寫，再拆成保留重複的 list；接著用 set 去除重複，並用交集、聯集和兩個方向的差集比較兩場的參加者；最後用 dict 保存摘要，計算各場人數、總不重複人數、報名人次，以及各場應收和總應收（桌遊每人 120 元、電影每人 80 元）。

本作業使用 tuple 保存活動名稱、dict 查詢費用並保存摘要、set 去重與比較名單，共完成三組測試。

## 測試一：原始資料
`movie_raw = " ben, Dora, ERIC, ben "`

```python
# HW09 兩場活動報名整理器
# ===== 輸入資料（測試一）=====

# tuple：保存固定的兩個活動名稱
activity_names = ("桌遊體驗", "電影欣賞")

# str：原始文字名單，保留原值不覆蓋
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# dict：用活動名稱查詢每人費用
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# ===== 1. 讀取活動名稱，整理成 list（保留順序與重複）=====

# 用拆包從 tuple 取出兩個活動名稱
board_name, movie_name = activity_names

# 先移除半形空白、再轉小寫、最後用逗號拆開
# 一定要先統一大小寫，"Amy" 和 "amy" 才會被視為同一人
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# ===== 2. 用 set 去重，並做集合運算 =====

# set 會自動去除重複，同一人在同一場只留一次
board_set = set(board_list)
movie_set = set(movie_list)

common_set = board_set & movie_set       # 交集：兩場都參加
all_set = board_set | movie_set          # 聯集：至少參加一場（同一人只算一次）
board_only_set = board_set - movie_set   # 差集：桌遊有、電影沒有
movie_only_set = movie_set - board_set   # 差集：電影有、桌遊沒有（方向相反，結果不同）

# ===== 3. 排序成 list，用切片取前兩位 =====

# set 沒有順序也不能用索引，先用 sorted() 轉成排序好的 list
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)
common_sorted = sorted(common_set)
board_only_sorted = sorted(board_only_set)
movie_only_sorted = sorted(movie_only_set)
all_sorted = sorted(all_set)

# 通知名單預覽：只取前兩位，不代表只有兩人能參加
notice_preview = all_sorted[:2]

# ===== 4. 用 dict 保存摘要 =====

summary = {
    "board_count": len(board_set),        # 桌遊人數（去重後）
    "movie_count": len(movie_set),        # 電影人數（去重後）
    "common_count": len(common_set),      # 共同人數
    "unique_count": len(all_set),         # 總不重複人數：一人只算一次
    "signup_times": len(board_set) + len(movie_set),  # 報名人次：參加兩場算兩人次
}

# 費用：用去重後人數 × 從 unit_fees 查到的單價（不用含重複的 list 長度）
summary["board_fee"] = summary["board_count"] * unit_fees[board_name]
summary["movie_fee"] = summary["movie_count"] * unit_fees[movie_name]
summary["total_fee"] = summary["board_fee"] + summary["movie_fee"]

# ===== 5. 輸出摘要（金額在輸出時才格式化到小數兩位）=====

print("HW09 活動報名摘要")
print(f"活動：{board_name}，每人 {unit_fees[board_name]:.2f} 元")
print(f"活動：{movie_name}，每人 {unit_fees[movie_name]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_sorted}")
print(f"電影去重名單：{movie_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{common_sorted}")
print(f"共同人數：{summary['common_count']} 人")
print(f"只參加桌遊者：{board_only_sorted}")
print(f"只參加電影者：{movie_only_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['unique_count']} 人")
print(f"報名人次：{summary['signup_times']} 人次")
print(f"通知名單前兩位：{notice_preview}")
print(f"桌遊應收：{summary['board_fee']:.2f} 元")
print(f"電影應收：{summary['movie_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
```
